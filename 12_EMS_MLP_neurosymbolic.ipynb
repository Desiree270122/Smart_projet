```python
# Cellule 2 : recharger le LSTM neurosymbolique deja entraine (pas de re-entrainement)
class LSTMNeuroSymbolique(nn.Module):
    def __init__(self, n_features, hidden_size, num_layers, dropout):
        super().__init__()
        self.lstm = nn.LSTM(input_size=n_features, hidden_size=hidden_size, num_layers=num_layers,
                             batch_first=True, dropout=dropout if num_layers > 1 else 0.0)
        self.head = nn.Sequential(nn.Linear(hidden_size, 32), nn.ReLU(), nn.Linear(32, 3))
    def forward(self, x):
        out, _ = self.lstm(x)
        return self.head(out[:, -1, :])

lstm_ns_scalers = np.load(MODELS_DIR / 'EMS_LSTM_neurosymbolic_scalers.npz', allow_pickle=True)
lstm_x_mean, lstm_x_std = lstm_ns_scalers['x_mean'], lstm_ns_scalers['x_std']
lstm_y_mean, lstm_y_std = lstm_ns_scalers['y_mean'], lstm_ns_scalers['y_std']

try:
    raw_state_dict = torch.load(LSTM_NS_CHECKPOINT, map_location=DEVICE, weights_only=True)
except TypeError:
    raw_state_dict = torch.load(LSTM_NS_CHECKPOINT, map_location=DEVICE)

n_features_checkpoint = raw_state_dict['lstm.weight_ih_l0'].shape[1]
hidden_size_checkpoint = raw_state_dict['lstm.weight_ih_l0'].shape[0] // 4

print('Architecture detectee dans le checkpoint :')
print('  nb features entree :', n_features_checkpoint)
print('  hidden_size        :', hidden_size_checkpoint)
print('Config LSTM_NS_INPUT_COLS attendue :', len(LSTM_NS_INPUT_COLS), '->', LSTM_NS_INPUT_COLS)

if n_features_checkpoint != len(LSTM_NS_INPUT_COLS):
    print(f"ATTENTION : le checkpoint attend {n_features_checkpoint} features, "
          f"LSTM_NS_INPUT_COLS en a {len(LSTM_NS_INPUT_COLS)}. "
          "Remplace LSTM_NS_INPUT_COLS_ACTUAL ci-dessous par la vraie liste de colonnes "
          "utilisee dans 08_lstm_neurosymbolique.ipynb (meme ordre que l'entrainement).")

LSTM_NS_INPUT_COLS_ACTUAL = LSTM_NS_INPUT_COLS  # <-- remplace ici si le message ci-dessus apparait

lstm_ns_model = LSTMNeuroSymbolique(n_features_checkpoint, hidden_size_checkpoint, LSTM_NS_NUM_LAYERS, LSTM_NS_DROPOUT).to(DEVICE)
lstm_ns_model.load_state_dict(raw_state_dict)
lstm_ns_model.eval()
print('LSTM-NS recharge depuis', LSTM_NS_CHECKPOINT)
```

```python
# Cellule 3 : generer les predictions du LSTM-NS pour chaque split (necessaire comme entree du MLP-NS)
def lstm_ns_predictions(data, window, horizon):
    x = data[lstm_input_cols_actual].to_numpy(dtype=np.float32)
    anchors = np.arange(window - 1, len(data) - horizon)
    Xw = np.empty((len(anchors), window, x.shape[1]), dtype=np.float32)
    for k, a in enumerate(anchors):
        Xw[k] = x[a - window + 1:a + 1]
    Xn = ((Xw - lstm_x_mean) / lstm_x_std).astype(np.float32)

    preds = []
    with torch.no_grad():
        for i in range(0, len(Xn), 512):
            batch = torch.from_numpy(Xn[i:i + 512]).to(DEVICE)
            preds.append(lstm_ns_model(batch).cpu().numpy())
    preds = np.concatenate(preds, axis=0) * lstm_y_std + lstm_y_mean

    pred_df = pd.DataFrame({
        'row_index': anchors,
        'Pdem_pred_ns': preds[:, 0],
        'delta_soc_eb_pred_ns': preds[:, 1],
        'delta_soc_pb_pred_ns': preds[:, 2],
    })
    return pred_df

lstm_preds = {name: lstm_ns_predictions(optimized_datasets[name], WINDOW, HORIZON) for name in ['train', 'validation', 'test']}
for name, p in lstm_preds.items():
    print(name, p.shape)
```

```python
# Cellule 4 : fusionner les predictions LSTM-NS dans les datasets (par position de ligne)
merged_datasets = {}
for name, data in optimized_datasets.items():
    d = data.reset_index(drop=True).copy()
    d['row_index'] = d.index
    merged = d.merge(lstm_preds[name], on='row_index', how='left')
    merged_datasets[name] = merged
    print(name, '-- lignes avec predictions LSTM-NS:', merged['Pdem_pred_ns'].notna().sum(), '/', len(merged))
```

```python
# Cellule 5 : construction X/y (alpha_star defini ET predictions LSTM-NS disponibles)
def build_xy_ns(data, cols, target_col):
    valid = data[target_col].notna() & data['alpha_ems_alpha_star_valid'].eq(1) & data['Pdem_pred_ns'].notna()
    X = data.loc[valid, cols].to_numpy(dtype=np.float32)
    y = data.loc[valid, target_col].to_numpy(dtype=np.float32)
    alpha_fuzzy_v = data.loc[valid, 'alpha_ems_fuzzy_logic'].to_numpy(dtype=np.float32)
    p_dem_v = data.loc[valid, 'hasPower'].to_numpy(dtype=np.float32)
    soc_eb_v = data.loc[valid, 'SOC_EB'].to_numpy(dtype=np.float32)
    soc_pb_v = data.loc[valid, 'SOC_PB'].to_numpy(dtype=np.float32)
    eb_avail_v = data.loc[valid, 'EB_available'].to_numpy(dtype=np.float32)
    return X, y, alpha_fuzzy_v, p_dem_v, soc_eb_v, soc_pb_v, eb_avail_v

(X_train, y_train, fuzzy_train, p_train, soceb_train, socpb_train, ebavail_train) = build_xy_ns(merged_datasets['train'], MLP_NS_INPUT_COLS, MLP_NS_TARGET)
(X_val, y_val, fuzzy_val, p_val, soceb_val, socpb_val, ebavail_val) = build_xy_ns(merged_datasets['validation'], MLP_NS_INPUT_COLS, MLP_NS_TARGET)
(X_test, y_test, fuzzy_test, p_test, soceb_test, socpb_test, ebavail_test) = build_xy_ns(merged_datasets['test'], MLP_NS_INPUT_COLS, MLP_NS_TARGET)

print('Train :', X_train.shape, '| Val :', X_val.shape, '| Test :', X_test.shape)
```

```python
# Cellule 6 : normalisation (les indicateurs binaires de MLP_NS_INPUT_COLS ne sont pas normalises)
BINARY_COLS_NS = ['EB_available', 'PB_available', 'EB_low_SOC', 'PB_low_SOC',
                   'high_power_demand', 'regenerative_braking', 'zero_power_demand', 'converter_risk']
is_binary = np.array([c in BINARY_COLS_NS for c in MLP_NS_INPUT_COLS])

x_mean_full = X_train.mean(axis=0)
x_std_full = X_train.std(axis=0)
x_mean = np.where(is_binary, 0.0, x_mean_full).astype(np.float32)
x_std = np.where(is_binary, 1.0, x_std_full).astype(np.float32)

if np.any(x_std[~is_binary] < 1e-12):
    bad = np.array(MLP_NS_INPUT_COLS)[~is_binary][x_std[~is_binary] < 1e-12]
    raise ValueError(f"Variables continues constantes dans le train : {bad.tolist()}")

def norm_x(X): return ((X - x_mean) / x_std).astype(np.float32)
Xn_train, Xn_val, Xn_test = norm_x(X_train), norm_x(X_val), norm_x(X_test)

MLP_NS_SCALER_FILE = MODELS_DIR / 'mlp_ns_scalers.npz'
np.savez(MLP_NS_SCALER_FILE, x_mean=x_mean, x_std=x_std, input_cols=np.array(MLP_NS_INPUT_COLS))
```

```python
# Cellule 7 : Dataset et DataLoader (X, alpha_fuzzy, Pdem, SOC_EB, SOC_PB, EB_available, cible alpha_star)
class NSDataset(Dataset):
    def __init__(self, X, alpha_fuzzy, p_dem, soc_eb, soc_pb, eb_avail, y):
        self.X = torch.from_numpy(X)
        self.alpha_fuzzy = torch.from_numpy(alpha_fuzzy)
        self.p_dem = torch.from_numpy(p_dem)
        self.soc_eb = torch.from_numpy(soc_eb)
        self.soc_pb = torch.from_numpy(soc_pb)
        self.eb_avail = torch.from_numpy(eb_avail)
        self.y = torch.from_numpy(y)
    def __len__(self): return len(self.X)
    def __getitem__(self, i):
        return (self.X[i], self.alpha_fuzzy[i], self.p_dem[i], self.soc_eb[i],
                self.soc_pb[i], self.eb_avail[i], self.y[i])

generator = torch.Generator()
generator.manual_seed(RANDOM_SEED)

train_ds = NSDataset(Xn_train, fuzzy_train, p_train, soceb_train, socpb_train, ebavail_train, y_train)
val_ds = NSDataset(Xn_val, fuzzy_val, p_val, soceb_val, socpb_val, ebavail_val, y_val)
test_ds = NSDataset(Xn_test, fuzzy_test, p_test, soceb_test, socpb_test, ebavail_test, y_test)

train_loader = DataLoader(train_ds, batch_size=MLP_NS_BATCH_SIZE, shuffle=True, generator=generator)
val_loader = DataLoader(val_ds, batch_size=MLP_NS_BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_ds, batch_size=MLP_NS_BATCH_SIZE, shuffle=False)
print(len(train_ds), len(val_ds), len(test_ds))
```

```python
# Cellule 8 : modele -- predit une correction bornee Delta_alpha appliquee a alpha_fuzzy
class MLPNeuroSymbolique(nn.Module):
    def __init__(self, n_in, hidden1, hidden2, dropout, max_delta):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(n_in, hidden1), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(hidden1, hidden2), nn.ReLU(),
            nn.Linear(hidden2, 1),
        )
        self.max_delta = max_delta

    def forward(self, x, alpha_ems_fuzzy_logic):
        z = self.net(x).squeeze(-1)
        delta_alpha = self.max_delta * torch.tanh(z)
        alpha = torch.clamp(alpha_ems_fuzzy_logic + delta_alpha, 0.0, 1.0)
        return alpha, delta_alpha

torch.manual_seed(RANDOM_SEED)
model = MLPNeuroSymbolique(len(MLP_NS_INPUT_COLS), MLP_NS_HIDDEN_1, MLP_NS_HIDDEN_2, MLP_NS_DROPOUT, MAX_DELTA_ALPHA).to(DEVICE)
print(model)
```

```python
# Cellule 9 : perte composite -- reutilise les composantes validees de 10_generation_alpha_optimal (en torch)
def cost_components_torch(alpha, p_dem, soc_eb, soc_pb, alpha_prev):
    p_pb = alpha * p_dem
    p_eb = p_dem - p_pb
    p_conv = torch.where(p_eb >= 0, p_eb / CONVERTER_RATIO_DISCHARGE, p_eb / CONVERTER_RATIO_CHARGE)

    eb_dir = torch.where(p_eb >= 0, torch.tensor(P_EB_MAX_W), torch.tensor(abs(P_EB_MIN_W)))
    pb_dir = torch.where(p_pb >= 0, torch.tensor(P_PB_MAX_W), torch.tensor(abs(P_PB_MIN_W)))
    conv_dir = torch.where(p_conv >= 0, torch.tensor(P_CONV_MAX_W), torch.tensor(abs(P_CONV_MIN_W)))
    eb_u = p_eb.abs() / eb_dir
    pb_u = p_pb.abs() / pb_dir
    conv_u = p_conv.abs() / conv_dir

    balance_cost = 0.5 * (eb_u ** 2 + pb_u ** 2)
    converter_cost = conv_u ** 2

    eb_soc_level = ((soc_eb - SOC_EB_MIN) / (SOC_EB_MAX - SOC_EB_MIN)).clamp(0, 1)
    pb_soc_level = ((soc_pb - SOC_PB_MIN) / (SOC_PB_MAX - SOC_PB_MIN)).clamp(0, 1)
    zero = torch.zeros_like(eb_soc_level)
    eb_risk = torch.where(p_dem > EPS_POWER_W, 1 - eb_soc_level, torch.where(p_dem < -EPS_POWER_W, eb_soc_level, zero))
    pb_risk = torch.where(p_dem > EPS_POWER_W, 1 - pb_soc_level, torch.where(p_dem < -EPS_POWER_W, pb_soc_level, zero))
    physics_cost = (1 - alpha) * eb_risk + alpha * pb_risk

    if alpha_prev is not None:
        continuity_cost = (alpha - alpha_prev) ** 2
    else:
        continuity_cost = torch.zeros_like(alpha)

    return balance_cost, physics_cost, converter_cost, continuity_cost


def composite_loss(alpha_pred, alpha_target, p_dem, soc_eb, soc_pb, eb_avail, alpha_prev):
    target_loss = ((alpha_pred - alpha_target) ** 2).mean()

    balance_cost, physics_cost, converter_cost, continuity_cost = cost_components_torch(
        alpha_pred, p_dem, soc_eb, soc_pb, alpha_prev
    )

    # L_regles : si EB protege (SOC_EB<=seuil) et traction, alpha doit etre proche de 1
    eb_protected = (1.0 - eb_avail)
    traction = (p_dem > EPS_POWER_W).float()
    rules_cost = eb_protected * traction * (1.0 - alpha_pred) ** 2

    total = (
        MLP_NS_LOSS_WEIGHTS['target'] * target_loss
        + MLP_NS_LOSS_WEIGHTS['rules'] * rules_cost.mean()
        + MLP_NS_LOSS_WEIGHTS['physics'] * physics_cost.mean()
        + MLP_NS_LOSS_WEIGHTS['balance'] * balance_cost.mean()
        + MLP_NS_LOSS_WEIGHTS['continuity'] * continuity_cost.mean()
        + MLP_NS_LOSS_WEIGHTS['converter'] * converter_cost.mean()
    )

    return total, {
        'target': target_loss.item(), 'rules': rules_cost.mean().item(),
        'physics': physics_cost.mean().item(), 'balance': balance_cost.mean().item(),
        'continuity': continuity_cost.mean().item(), 'converter': converter_cost.mean().item(),
    }
```

```python
# Cellule 10 : entrainement (alpha_prev = la cible alpha_star precedente, connue pendant l'entrainement)
optimizer = torch.optim.Adam(model.parameters(), lr=MLP_NS_LEARNING_RATE)
best_val = float('inf')
best_epoch = -1
patience_count = 0
history = {'train': [], 'val': []}

for epoch in range(MLP_NS_EPOCHS):
    model.train()
    train_loss = 0.0
    comp_sum = {k: 0.0 for k in ['target', 'rules', 'physics', 'balance', 'continuity', 'converter']}
    for xb, fz, pd_, se, sp, ea, yb in train_loader:
        xb, fz, pd_, se, sp, ea, yb = [t.to(DEVICE) for t in (xb, fz, pd_, se, sp, ea, yb)]
        optimizer.zero_grad()
        alpha_pred, _ = model(xb, fz)
        alpha_prev = torch.cat([yb[:1] * 0 + yb[0], yb[:-1]])  # approx: cible precedente dans le batch
        loss, comps = composite_loss(alpha_pred, yb, pd_, se, sp, ea, alpha_prev)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        train_loss += loss.item() * xb.size(0)
        for k in comp_sum: comp_sum[k] += comps[k] * xb.size(0)
    train_loss /= len(train_ds)
    for k in comp_sum: comp_sum[k] /= len(train_ds)

    model.eval()
    val_loss = 0.0
    with torch.no_grad():
        for xb, fz, pd_, se, sp, ea, yb in val_loader:
            xb, fz, pd_, se, sp, ea, yb = [t.to(DEVICE) for t in (xb, fz, pd_, se, sp, ea, yb)]
            alpha_pred, _ = model(xb, fz)
            alpha_prev = torch.cat([yb[:1], yb[:-1]])
            loss, _ = composite_loss(alpha_pred, yb, pd_, se, sp, ea, alpha_prev)
            val_loss += loss.item() * xb.size(0)
    val_loss /= len(val_ds)

    history['train'].append(train_loss)
    history['val'].append(val_loss)

    if not np.isfinite(train_loss) or not np.isfinite(val_loss):
        raise ValueError("Loss non finie detectee.")

    if val_loss < best_val - 1e-6:
        best_val, best_epoch, patience_count = val_loss, epoch, 0
        torch.save(model.state_dict(), MLP_NS_CHECKPOINT)
    else:
        patience_count += 1

    if epoch == 0 or (epoch + 1) % 5 == 0:
        print(f"Epoch {epoch+1:3d}/{MLP_NS_EPOCHS} train={train_loss:.5f} val={val_loss:.5f} "
              f"[target={comp_sum['target']:.4f} rules={comp_sum['rules']:.4f} physics={comp_sum['physics']:.4f} "
              f"balance={comp_sum['balance']:.4f} cont={comp_sum['continuity']:.4f} conv={comp_sum['converter']:.4f}] "
              f"patience={patience_count}")

    if patience_count >= MLP_NS_PATIENCE:
        print(f"Arret anticipe a l'epoque {epoch+1}.")
        break

try:
    state_dict = torch.load(MLP_NS_CHECKPOINT, map_location=DEVICE, weights_only=True)
except TypeError:
    state_dict = torch.load(MLP_NS_CHECKPOINT, map_location=DEVICE)
model.load_state_dict(state_dict)
print('Meilleure epoque :', best_epoch + 1, '| val :', best_val)
```
